```python
from pyspark.sql import functions as F

s3_path = "s3://ujjivanpoc/source/bank_transactions/Bank_Transaction_Fraud_Detection.csv"

df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(s3_path)
)

# Preview / test with 10 records first
df_sample = df.limit(10)
df_sample.show(truncate=False)

# Once validated, write the sample (or swap df_sample -> df for full load)
(
    df_sample.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("ujjivan_2.bronze.bank_transaction_fraud_detection_3")
)
```
